```python
using DifferentialEquationsusing Plots

# Parameters
k_plus = 1e6    # M⁻¹ s⁻¹
k_minus = 10    # s⁻¹
R_total = 1e-6  # μM
L_init = 1e-6   # μM

# Initial conditions: [RL]
u0 = [0.0]

# Time span
tspan = (0.0, 10.0)

# ODE function
function binding!(du, u, p, t)
    RL = u[1]
    R_free = R_total - RL
    L_free = L_init - RL
    du[1] = k_plus * R_free * L_free - k_minus * RL
end

prob = ODEProblem(binding!, u0, tspan)
sol = solve(prob)

fraction_bound = sol[1,:] ./ R_total

plot(sol.t, fraction_bound, label="Bound fraction", xlabel="Time (s)", ylabel="Fraction bound", title="Ligand-Receptor Binding (Kd = $(k_minus/k_plus) M)")
```
