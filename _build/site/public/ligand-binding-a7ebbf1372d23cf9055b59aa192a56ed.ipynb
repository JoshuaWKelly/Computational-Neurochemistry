```python
import nglview as nv
from nglview.datafiles import PDB

# Load a structure (e.g., from RCSB PDB or local file)
view = nv.show_pdbid("6OY9")  # Example: Opioid receptor
# Or local: view = nv.show_file("my_protein.pdb")

# Customize representations
view.add_cartoon(color='spectrum')
view.add_surface(opacity=0.4, color='hydrophobicity')
view.add_licorice('ligand')  # Highlight bound ligand

# Plain-English description for accessibility
print("Interactive view: Cartoon representation colored by chain, semi-transparent surface by hydrophobicity.")

view  # Displays the widget
```
